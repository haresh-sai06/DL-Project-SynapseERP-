# Dukaan AI Evaluation

Reads saved metrics from training scripts and prepares the final result table.

```python
import json
from pathlib import Path
ROOT = Path.cwd().resolve().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
metrics_dir = ROOT / 'outputs' / 'metrics'
metrics = {}
for path in metrics_dir.glob('*.json'):
    metrics[path.stem] = json.loads(path.read_text())
metrics
```

```python
import pandas as pd
rows = []
for module, values in metrics.items():
    for metric, value in values.items():
        rows.append({'module': module, 'metric': metric, 'value': value})
pd.DataFrame(rows)
```
